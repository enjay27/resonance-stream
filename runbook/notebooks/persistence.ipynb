# Persistence — what is still there after a restart

Over the bridge, no clicks: chat is replayed in three channels (WORLD, which the archive ignores by default, GUILD and PARTY), the favorites popup's `save-favorites` is pressed (the block must not undo it), a sender is blocked, the app is
closed with `quit`; the notebook then reads the files on disk (the day's chat log, `config.json`), starts the app again on the **same data folder** and asks what it
serves: the archived lines in order, the blocked sender's rows still flagged, a new line numbered after the restored ones, and the favorites unchanged, tab ids included (`favorites-changed` is published once). It also checks the version the app announces (a stale exe says so), that `chat_log_retention_days` deletes old day logs at start-up and nothing else, and that a flood of WORLD lines does not push GUILD out of the reload. Needs a `test-env` exe and
Node 20+; run Jupyter as Administrator. The same pipeline without Jupyter: `python -m runbook.run persistence`.

```python
import sys
from pathlib import Path
here = Path.cwd().resolve()              # Jupyter starts a notebook in its own folder:
root = next(p for p in [here, *here.parents] if (p / "runbook" / "common.py").exists())
sys.path.insert(0, str(root))            # ...so find runbook above it
from runbook import bridge, common
from runbook.common import Recorder
from runbook.pipelines.persistence import Persistence
rec = Recorder("persistence")
if not common.require_windows_admin(rec):
    raise SystemExit("Start Jupyter from an Administrator terminal, then run again.")
bridge.ensure_installed()                # npm ci in runbook/bridge, the first time
```

```python
EXE = common.ask_exe()          # target/release/resonance-stream.exe of this checkout, unless RUNBOOK_TEXT_LOCAL_EXE says otherwise
Persistence(rec, EXE, common.RUNS).run()
```
