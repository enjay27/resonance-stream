# Popup windows — the cheat sheet and favorites, with real windows and no clicks

Over the bridge: both popups exist (hidden) a moment after start; `open-popup` shows one and a second open makes no second window; the X (`hide-popup`) hides it and the
next open shows it again; the pin on the overlay pins both popups; closing the main window ends the app and leaves no process; a popup left at a place (`place-popup`) opens
there after a restart. Needs a `test-env` exe and Node 20+; run Jupyter as Administrator. The same pipeline without Jupyter: `python -m runbook.run popups`.

```python
import sys
from pathlib import Path
here = Path.cwd().resolve()              # Jupyter starts a notebook in its own folder:
root = next(p for p in [here, *here.parents] if (p / "runbook" / "common.py").exists())
sys.path.insert(0, str(root))            # ...so find runbook above it
from runbook import bridge, common
from runbook.common import Recorder
from runbook.pipelines.popups import Popups
rec = Recorder("popups")
if not common.require_windows_admin(rec):
    raise SystemExit("Start Jupyter from an Administrator terminal, then run again.")
bridge.ensure_installed()                # npm ci in runbook/bridge, the first time
```

```python
EXE = common.ask_exe()          # target/release/resonance-stream.exe of this checkout, unless RUNBOOK_TEXT_LOCAL_EXE says otherwise
Popups(rec, EXE, common.RUNS).run()
```
