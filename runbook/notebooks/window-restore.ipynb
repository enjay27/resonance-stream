# K18 — closing with Settings open must not save the enlarged window

Settings makes the window bigger; the window-state plugin saves the size the window has at exit, so the next start would open enlarged
unless the app puts the old size back first. The notebook does what you did by hand, **over the bridge, with no clicks**: it grows the
window (what opening Settings does), leaves by **quit** (the tray's Quit) and again by **close** (the window's X), starts the app again on the
same data folder with the window state on, and compares the window's size with the one before the grow. Needs a `test-env` exe and Node 20+;
run Jupyter as Administrator. The same pipeline without Jupyter: `python -m runbook.run window-restore`.

```python
import sys
from pathlib import Path
here = Path.cwd().resolve()              # Jupyter starts a notebook in its own folder:
root = next(p for p in [here, *here.parents] if (p / "runbook" / "common.py").exists())
sys.path.insert(0, str(root))            # ...so find runbook above it
from runbook import bridge, common
from runbook.common import Recorder
from runbook.pipelines.window_restore import WindowRestore
rec = Recorder("window-restore")
if not common.require_windows_admin(rec):
    raise SystemExit("Start Jupyter from an Administrator terminal, then run again.")
bridge.ensure_installed()                # npm ci in runbook/bridge, the first time
```

```python
EXE = common.ask_exe()          # target/release/resonance-stream.exe of this checkout, unless RUNBOOK_TEXT_LOCAL_EXE says otherwise
WindowRestore(rec, EXE, common.RUNS).run()
```
